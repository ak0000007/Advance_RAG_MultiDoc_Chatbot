```python
chunker = text_chunker(
    chunk_size  =1000,
    chunk_overlap = 150
)

all_chunks  []

for document in documents:

    chunks = chunker.chunk_document(
        document
    )

    all_chunks.extend(
        chunks
    )


print(
    f"Created {len(all_chunks)} chunks"
)
```

```python
texts = [
    chunk.content
    for chunk in all_chunks
]


embeddings = embedder.embed_documents(
    texts
)


print(
    "Number of chunks:",
    len(all_chunks)
)


print(
    "Embedding shape:",
    embeddings.shape
)
```

```python
dimension = embeddings.shape[1]


vector_store = FAISSVectorStore(
    dimension=dimension
)
dimension = embeddings.shape[1]
```

```python
vector_store.add(
    embeddings,
    all_chunks
)


print(
    "Vectors in FAISS:",
    vector_store.index.ntotal
)
```

```python
query = "What is the payment deadline?"
query_embedding = embedder.embed_query(
    query
)
results = vector_store.search(
    query_embedding,
    top_k=5
)
```

```python
for result in results:

    print(
        "Score:",
        result["score"]
    )

    print(
        "Document:",
        result["chunk"].metadata.get(
            "document_name"
        )
    )

    print(
        "Content:"
    )

    print(
        result["chunk"].content[:500]
    )

    print(
        "-" * 80
    )
```
